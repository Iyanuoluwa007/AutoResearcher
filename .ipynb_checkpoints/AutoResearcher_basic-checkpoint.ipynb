## 2) Load and chunk PDFs
We use LangChain's `PyPDFLoader` and a `RecursiveCharacterTextSplitter` to make chunks for retrieval.

> If there are no PDFs in `data/`, the loader will warn and skip.

```python
# %pip install pypdf
```

```python
import glob
from langchain_community.document_loaders import PyPDFLoader
from langchain_text_splitters import RecursiveCharacterTextSplitter
from langchain_core.documents import Document

DATA_DIR = 'data'
pdf_paths = sorted(glob.glob(os.path.join(DATA_DIR, '*.pdf')))
if not pdf_paths:
    print('⚠️ No PDFs found in ./data. Add a few PDFs (YOLO/ROS2/SLAM etc.) and re-run this cell.')

docs = []
for p in pdf_paths:
    try:
        for d in PyPDFLoader(p).load():
            d.metadata = (d.metadata or {}) | {'source': p}
            docs.append(d)
    except Exception as e:
        print(f'Failed to load {p}: {e}')

print(f'Loaded documents: {len(docs)}')

splitter = RecursiveCharacterTextSplitter(chunk_size=1000, chunk_overlap=150)
chunks = splitter.split_documents(docs) if docs else []
print(f'Chunks: {len(chunks)}')
```

## 3) Build / Load Chroma index
We use OpenAI embeddings by default (high quality). If you want to go offline, swap to `SentenceTransformerEmbeddings` later.

```python
# %pip install -q sentence-transformers
```

```python
from langchain_community.vectorstores import Chroma
# ✅ Use offline embeddings (no OpenAI key/cost)
from langchain_community.embeddings import SentenceTransformerEmbeddings
import os

INDEX_DIR = 'data/chroma_local_offline'  # use a new folder to avoid mixed indexes
os.makedirs(INDEX_DIR, exist_ok=True)

# Use a strong, lightweight model. Good defaults:
# "all-MiniLM-L6-v2" (fast, small) or "all-mpnet-base-v2" (better quality, heavier)
EMBED_MODEL_NAME = "all-MiniLM-L6-v2"
emb = SentenceTransformerEmbeddings(model_name=EMBED_MODEL_NAME)

def build_index_offline(_chunks):
    if not _chunks:
        print('⚠️ No chunks to index yet. Add PDFs to ./data and re-run loading.')
        return None
    vs = Chroma.from_documents(
        documents=_chunks,
        embedding=emb,
        persist_directory=INDEX_DIR,
    )
    vs.persist()
    return vs

def load_index_offline():
    try:
        return Chroma(persist_directory=INDEX_DIR, embedding_function=emb)
    except Exception as e:
        print('Could not load existing index:', e)
        return None

# Build if we have chunks; otherwise try loading an existing index
vs = build_index_offline(chunks) if chunks else load_index_offline()
print('Vector store ready (offline):', bool(vs), '| model:', EMBED_MODEL_NAME)
```

## 4) Minimal LLM wrapper (OpenAI)
We use the official OpenAI SDK and a tiny helper that takes a list of messages and returns the text. Make sure your `.env` has `OPENAI_API_KEY`.
If you don't have an API key yet, you can still run the retrieval part; the final synthesis step requires a model.

```python
import os
from openai import OpenAI

USE_OPENAI = os.getenv('USE_OPENAI', 'true').lower() == 'true'
client = OpenAI() if USE_OPENAI else None

def chat(messages, model='gpt-4o-mini'):
    if not client:
        raise RuntimeError('OpenAI client not configured. Set OPENAI_API_KEY in .env or set USE_OPENAI=true.')
    resp = client.chat.completions.create(model=model, messages=messages)
    return resp.choices[0].message.content

print('LLM ready:', bool(client))
```

## 5) Retrieve → Synthesize function
This is the core of RAG: we search the vector DB, pass the snippets into a prompt, and ask the model to synthesize a **well-cited answer**.

```python
from typing import List, Dict

def retrieve(query: str, k: int = 6) -> List[Dict]:
    if not vs:
        return []
    hits = vs.similarity_search(query, k=k)
    out = []
    for h in hits:
        out.append({
            'content': h.page_content,
            'source': (h.metadata or {}).get('source', 'local')
        })
    return out

def synthesize_answer(question: str, retrieved: List[Dict], word_budget: int = 300) -> str:
    if not retrieved:
        return 'No retrieved context yet. Add PDFs to ./data and rebuild the index.'
    context = '\n\n'.join([f"[{i}] ({r['source']}) {r['content'][:1200]}" for i,r in enumerate(retrieved)])
    prompt = f"""Synthesize a concise, well-cited answer to the user's question using only the context snippets.
Question: {question}
Context:
{context}

Return sections:
1) Key findings
2) Comparison table (if applicable)
3) Cited references [use bracket index numbers]
4) Next actions for the user (practical steps)
Keep it < {word_budget} words.
"""
    return chat([{"role": "user", "content": prompt}])
```

- *"List key ROS 2 perception stacks suitable for embedded robots; compare and cite."*

```python
# Install requests if you don't have it already
# %pip install -q requests
```

```python
# Hybrid chat(): try OpenAI first, fall back to local Ollama if 429/No key/etc.
import os, requests
from typing import List, Dict

# Load environment variables from .env file
from dotenv import load_dotenv
load_dotenv()

USE_OPENAI = os.getenv("USE_OPENAI", "true").lower() == "true"
OPENAI_OK = bool(os.getenv("OPENAI_API_KEY"))

OLLAMA_URL = "http://localhost:11434/api/generate"
OLLAMA_MODEL = "llama3.2"  # change to qwen2.5:7b-instruct or mistral:7b-instruct if you prefer

def chat(messages: List[Dict[str, str]], model: str = "gpt-4o-mini"):
    """
    Try OpenAI first (if key present), else fall back to Ollama.
    """
    # 1) Try OpenAI if possible
    if USE_OPENAI and OPENAI_OK:
        try:
            from openai import OpenAI
            client = OpenAI()
            resp = client.chat.completions.create(model=model, messages=messages,
                                                  max_tokens=500)  # keep modest
            return resp.choices[0].message.content
        except Exception as e:
            # If rate-limited or any other error, fall back to Ollama
            print("OpenAI failed, falling back to Ollama ->", e)

    # 2) Ollama fallback (local, free)
    prompt = "\n".join([f"{m['role'].upper()}: {m['content']}" for m in messages])
    r = requests.post(OLLAMA_URL, json={"model": OLLAMA_MODEL, "prompt": prompt, "stream": False}, timeout=120)
    r.raise_for_status()
    data = r.json()
    return data.get("response", "").strip()
```

```python
QUESTION = "Summarize YOLOv11 vs YOLOv8 differences for robotics vision on embedded devices"
retrieved = retrieve(QUESTION, k=6)
print(f"Retrieved {len(retrieved)} chunks.")
if retrieved:
    print('Sources:')
    for i, r in enumerate(retrieved):
        print(f"[{i}] {r['source']}")

answer = synthesize_answer(QUESTION, retrieved, word_budget=300)
print('\n---\n')
print(answer)
```
